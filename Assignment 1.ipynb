```python
# Lists
number = [10,20,30,40]
```

```python
number = [10,20,30,40]
#lists operation append()
number.append(50)
print(number)
```

```python
number = [10,20,30,40]
#lists operation insert()
number.insert(0,5)
print(number)
```

```python
number = [10,20,30,40]
#lists operation remove()
number.remove(30)
print(number)
```

```python
number = [10,20,30,40]
#lists operation pop()
number.pop()
print(number)
```

```python
number = [10,20,30,40]
#lists operation sort()
number.sort()
print(number)
```
